> Note *ignoring any other text on the line*: version 4 also outputs quoted evidence, so the scorer has to cope with lines that are not just answers. The reference cell below already handles that.

````python
import re, matplotlib.pyplot as plt

# Paste each set of answers between the triple quotes, exactly as the model
# gave them. Extra lines, quotes and commentary are ignored by the parser.
v1 = '''

'''

v2 = '''

'''

v3 = '''

'''

v4 = '''

'''

truth = dict(zip(gold['id'], gold['correct_date']))

PATTERN = r'(L\d+)\s*[:\-]\s*([0-9X]{4}-[0-9X]{2}-[0-9X]{2}|NOT STATED)'

def parse(blob):
    out = {}
    for line in blob.splitlines():
        m = re.search(PATTERN, line, re.I)
        if m:
            out[m.group(1)] = m.group(2).strip().upper()
    return out

def score(blob):
    got = parse(blob)
    if not got:
        return 0.0, []   # nothing pasted yet
    right = [i for i in truth if got.get(i, '') == truth[i].upper()]
    return len(right) / len(truth), right

results = {}
for name, v in [('v1 naive', v1), ('v2 +form', v2),
                ('v3 +examples', v3), ('v4 +evidence', v4)]:
    acc, right = score(v)
    results[name] = acc
    print(f'{name:16s} {acc:6.1%}')

if any(results.values()):
    plt.figure(figsize=(6.5, 3.5))
    plt.bar(results.keys(), results.values(),
            color=['#CFC5B7', '#B84A5C', '#8E2436', '#12706A'])
    plt.ylabel('accuracy'); plt.ylim(0, 1)
    plt.xticks(rotation=15); plt.tight_layout(); plt.show()

    _, r1 = score(v1)
    _, r4 = score(v4)
    print('\nFixed by the better prompt:')
    for i in sorted(set(r4) - set(r1)):
        print(f'  {i}: correct answer {truth[i]}')
````
