## 1 · Get the text out

The unglamorous step that decides whether any of this works. A PDF of *images* — a photographed manuscript, an old scan — contains no text at all, and everything downstream will then silently operate on nothing.

### 1 · Extract, and check what you got

**Copy this into the AI box** — click the word **generate** in the empty cell below, then paste this in:

```text
Download https://fabsilvestri.github.io/ai-from-the-inside/corpus.zip
and unzip it into the current folder, so that letters.csv and the
archive/ and pdfs/ folders are alongside this notebook.

Install pypdf if it is not already available.

Read every PDF in the pdfs folder and every .txt file in the archive folder.
Build a dataframe called `docs` with one row per file and these columns:
filename, pages, chars, and text (the full extracted text).

If a file yields no text at all, record 'NO TEXT LAYER' for it and carry on
with the others - do not stop the whole run.
Show the filename, pages and chars columns sorted by chars ascending.
```

> **Read that table before going any further.** Sorting ascending puts the failures at the top where you cannot miss them. A file with 0 characters is invisible to everything that follows. Note the prompt asks for the text to be *kept* — a table of statistics with the text thrown away would leave section 2 with nothing to work on.

<details><summary><b>Reference — click to open, or just run the cell below</b></summary>

This is one correct answer. Colab's AI will probably write something different, and that is fine — compare the two.

</details>

````python
!pip -q install pypdf
import pathlib, pandas as pd
from pypdf import PdfReader

records = []
for p in sorted(pathlib.Path('pdfs').glob('*.pdf')):
    try:
        r = PdfReader(str(p))
        text = '\n'.join((pg.extract_text() or '') for pg in r.pages)
        records.append({'file': p.name, 'pages': len(r.pages),
                        'chars': len(text.strip()), 'text': text})
    except Exception as e:
        records.append({'file': p.name, 'pages': 0, 'chars': 0, 'text': '',
                        'note': f'FAILED: {e}'})

for p in sorted(pathlib.Path('archive').glob('*.txt')):
    text = p.read_text(encoding='utf-8')
    records.append({'file': p.name, 'pages': 1, 'chars': len(text.strip()),
                    'text': text})

docs = pd.DataFrame(records)
docs['status'] = docs['chars'].apply(lambda c: 'NO TEXT LAYER' if c == 0 else 'ok')
# the six sample PDFs are the same letters as six of the .txt files
docs = docs.drop_duplicates(subset='text', keep='first').reset_index(drop=True)
print(f"{len(docs)} distinct documents, "
      f"{(docs.status != 'ok').sum()} with no usable text\n")
display(docs[['file', 'pages', 'chars', 'status']].sort_values('chars').head(12))
````

## 2 · Cut it into passages

A whole document is too big to put on the model's desk and too coarse to search. We cut it into passages, keeping a note of where each one came from.

Our letters are short — most will end up as a single passage each. On real material (a thesis chapter, an interview transcript) the splitting does real work.

### 2 · Chunk, keeping provenance

**Copy this into the AI box** — click the word **generate** in the empty cell below, then paste this in:

```text
Split the text of every document in `docs` into passages of about 120 words
with 30 words of overlap. Build a dataframe called `chunks` with columns:
n (a running number), file, part, and text.
Print how many passages there are, how many documents produced more than
one passage, and show the first two passages in full.
```

> Provenance is what makes a citation mean something later. Without it you get an answer with a number attached that you cannot trace back to anything.

<details><summary><b>Reference — click to open, or just run the cell below</b></summary>

This is one correct answer. Colab's AI will probably write something different, and that is fine — compare the two.

</details>

````python
def chunk(text, size=120, overlap=30):
    words = text.split()
    step = max(size - overlap, 1)
    return [' '.join(words[i:i + size]) for i in range(0, max(len(words), 1), step)
            if words[i:i + size]]

passages = []
for _, row in docs[docs.status == 'ok'].iterrows():
    for j, piece in enumerate(chunk(row['text'])):
        passages.append({'n': len(passages), 'file': row['file'],
                         'part': j, 'text': piece})

chunks = pd.DataFrame(passages)
multi = chunks.groupby('file').size().gt(1).sum()
print(f'{len(chunks)} passages from {chunks.file.nunique()} documents')
print(f'{multi} documents were long enough to split into more than one passage')
print('(our letters are short — on your own material this does much more work)\n')
for i in range(min(2, len(chunks))):
    r = chunks.iloc[i]
    print(f"--- passage {r['n']}  ({r['file']}, part {r['part']}) ---")
    print(r['text'][:400], '\n')
````

## 3 · Search

Now the interesting part, and the one where you should be sceptical.

We turn every passage into a list of numbers — its coordinates — using a classical method (TF-IDF followed by a dimensionality reduction, known as **LSA**). Passages that use words in similar patterns end up near each other. It needs nothing beyond what Colab already has.

This is a *weaker relative* of the embeddings from this morning. Watching exactly where it succeeds and where it fails is the point of this section.

### 3 · Build the search, and test it on its own

**Copy this into the AI box** — click the word **generate** in the empty cell below, then paste this in:

```text
Turn every passage in `chunks` into a vector using TfidfVectorizer with
English stop words and bigrams, then TruncatedSVD with 100 components,
then normalise. Keep the fitted pipeline as `embedder` and the matrix as `V`.

Write a function search(question, k=5) that returns the k most similar
passages by cosine similarity, each with its similarity score, passage
number and source filename.

Test it with the question 'someone jailed with no accusation against them'
and print the results. Warn me if the best score is below 0.15.
```

> **Test retrieval before you connect it to anything.** If the search is bad, the answers will be confidently wrong and you will blame the wrong component.

<details><summary><b>Reference — click to open, or just run the cell below</b></summary>

This is one correct answer. Colab's AI will probably write something different, and that is fine — compare the two.

</details>

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
from sklearn.pipeline import make_pipeline
import numpy as np

embedder = make_pipeline(
    TfidfVectorizer(stop_words='english', ngram_range=(1, 2), min_df=2),
    TruncatedSVD(n_components=min(100, len(chunks) - 1), random_state=0),
    Normalizer())
V = embedder.fit_transform(chunks['text'])

def search(question, k=5):
    q = embedder.transform([question])
    sims = (V @ q.T).ravel()
    top = np.argsort(sims)[::-1][:k]
    return [{'n': int(chunks.iloc[i]['n']), 'file': chunks.iloc[i]['file'],
             'score': float(sims[i]), 'text': chunks.iloc[i]['text']} for i in top]

def show_hits(question):
    hits = search(question)
    print(f'QUESTION: {question}\n')
    if hits[0]['score'] < 0.15:
        print('*** WARNING: nothing in the archive is close to this question.')
        print('*** The passages below are near-arbitrary. Do not trust an answer')
        print('*** built on them.\n')
    for h in hits:
        print(f"[{h['n']:3d}] {h['score']:.3f}  {h['file']}")
        print('     ', h['text'][:200], '...\n')

show_hits('someone jailed with no accusation against them')
````

---
## 4 · Now find out what it cannot do

Run the three questions below one at a time. They are chosen to show you three different behaviours — and the third one is the reason this section exists.

````python
# Ask anything you like — just change the text between the quotes.
show_hits('the river overflowed and destroyed property')
````

````python
show_hits('an outbreak of disease has ended')
````

````python
show_hits('Did cold weather ruin anyone\'s grapes?')
````

What you should have seen:

| Question | What happened | Why |
|---|---|---|
| *the river overflowed…* | Found the flood letters, though it never used the word *flood* | Those letters say **waters rose**, **broke its banks** and **flood** in the same documents, so the method could connect them |
| *an outbreak of disease…* | Score near zero, warning fires | The archive says *fever*, *contagion*, *plague* — it has never seen the words *outbreak* or *disease* at all |
| *cold weather… grapes* | Confidently returned the wrong letter | It matched the words **cold weather** in letters about building work |

That middle row is the honest limit of this method: **it can only relate words it has actually seen together.** Ask in vocabulary the archive does not contain and you get nothing — loudly, which is the good case.

The third row is the dangerous one, and it is worse than it looks. The archive **does** answer that question — several letters say *the frost took the lower rows of the vineyard* and *the vines suffered in the late cold*. Scroll the results and you will find one, somewhere around seventh place, at less than half the score of the building-works letter sitting at the top.

So this is not 'no answer exists'. This is **the right answer was there, and the wrong one beat it comfortably** — and nothing in the output says so. That is what silent failure looks like.

> Commercial systems do better here, because they use neural embeddings trained on enormous amounts of text — which need a GPU to build and to run. That is one of the reasons a university keeps its own machines, and where this course goes next.

> **The lesson that transfers:** bad retrieval is invisible in the answer. The only way to know is to look at what was retrieved.

---
## 5 · Put it together

Now assemble the desk: the retrieved passages, plus your question, plus the instruction that does the actual work.

### 5 · Build the prompt assembler

**Copy this into the AI box** — click the word **generate** in the empty cell below, then paste this in:

```text
Write a function ask(question) that:
  1. retrieves the 5 most relevant passages using search()
  2. prints a warning if the best score is below 0.15
  3. prints a single block of text containing, in this order: the
     instruction below, the numbered passages each with its source
     filename, and then the question

The instruction must read:
  Answer using ONLY the passages below. After every statement, cite the
  passage number it came from, like [3]. If the passages do not contain
  the answer, reply exactly: 'The provided passages do not answer this.'
  Do not use anything you know from outside these passages.
```

> Notice that you are specifying the architecture of a retrieval system in plain English. That is the whole skill.

<details><summary><b>Reference — click to open, or just run the cell below</b></summary>

This is one correct answer. Colab's AI will probably write something different, and that is fine — compare the two.

</details>

````python
INSTRUCTION = '''Answer using ONLY the passages below. After every statement,
cite the passage number it came from, like [3]. If the passages do not contain
the answer, reply exactly: "The provided passages do not answer this."
Do not use anything you know from outside these passages.'''

def ask(question, k=5):
    hits = search(question, k)
    if hits[0]['score'] < 0.15:
        print('*** WARNING: retrieval found nothing close. Expect a bad answer.\n')
    print('--- copy everything below into the Colab AI chat panel ---\n')
    print(INSTRUCTION)
    print('\nPASSAGES:')
    for h in hits:
        print(f"[{h['n']}] (from {h['file']}) {h['text']}\n")
    print(f'QUESTION: {question}')

ask('someone asking for a delay before they repay what they owe')
````

Change the question in the cell above and run it again — that is how you use it. Paste the block into the chat panel and read what comes back.

**Then open the passage it cites and check that the passage actually says what the answer claims.** That is the habit worth taking away.

---
## 6 · The trust test — do not skip this

Before trusting this with anything that matters, run these four checks.

1. Ask about **three things you know are in the archive**. Check both the answer and the passage it cited.
2. Ask about something you know is **absent** — try the cell below. Does it say so?
3. Ask about something present but **phrased differently** from the archive's own wording. Did retrieval still find it?
4. Ask something whose words appear in the archive but in the **wrong context** (like the grapes question). Did the answer notice?

Step 2 is the one everybody skips, and it is the one that matters. **A system that never admits ignorance is not a research tool.**

````python
ask('What did the Bishop say about the printing press in Amsterdam?')
````

Look at what came back: letters about imprisonment and unpaid debts, at scores around 0.43. Nothing whatsoever to do with printing presses in Amsterdam — and note that the score is **too high to trigger the warning**. The score guard will not save you here. The only thing standing between you and an invented answer is the *instruction*, and a well-instructed model will reply *'The provided passages do not answer this.'*

If yours invents an answer instead, your instruction is not strong enough. Go back to section 5, strengthen it, and test again. That loop — test, fail, strengthen, retest — is the actual work.

---
## Five honest failure modes

| Failure | How you catch it |
|---|---|
| Retrieval misses the right passage | Ask about something you know is there |
| Retrieval returns the wrong passage confidently | Read the passages, not just the answer |
| A PDF had no text layer | The extraction table in section 1 |
| The model blends in outside knowledge | Citations — check the passage really says it |
| **It cannot tell you what is absent** | Nothing catches this. It is a hard limit. |

That last row deserves a moment. *'No letter in this archive mentions the flood'* is a claim this architecture **can never justify** — it only ever looked at five passages, never at the archive. Arguments from absence need a different method.

> **Playbook line:** before trusting any grounded system, test it on facts you already know — including one it should not be able to answer.
